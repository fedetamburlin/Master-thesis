```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Impostazione del tema grafico per un aspetto pulito.
sns.set_theme(style="whitegrid")

# 2. Caricamento del dataset dal file CSV.
#    Il file deve trovarsi nella stessa directory di esecuzione dello script.
try:
    df = pd.read_csv('AR_DR_results.csv')

    # 3. Calcolo della matrice di correlazione.
    correlation_matrix = df.corr()

    # 4. Inizializzazione della figura e degli assi per il plot.
    #    Le dimensioni (figsize) sono aumentate per garantire la leggibilità.
    fig, ax = plt.subplots(figsize=(14, 12))

    # 5. Generazione della heatmap.
    #    - correlation_matrix: i dati da plottare.
    #    - annot=True: mostra i valori numerici delle correlazioni.
    #    - fmt=".2f": formatta i valori con due cifre decimali.
    #    - cmap='coolwarm': mappa di colori divergente, ideale per correlazioni.
    #    - square=True: impone una forma quadrata per le celle.
    #    - linewidths=.5: aggiunge una linea di separazione tra le celle.
    #    - annot_kws={"size": 10}: imposta la dimensione del font per le annotazioni.
    #    - cbar_kws={"shrink": .8}: riduce leggermente la dimensione della color bar.
    heatmap = sns.heatmap(
        correlation_matrix,
        annot=True,
        fmt=".2f",
        cmap='coolwarm',
        square=True,
        linewidths=.5,
        annot_kws={"size": 10},
        cbar_kws={"shrink": .8},
        ax=ax
    )

    # 6. Impostazione del titolo del grafico.
    ax.set_title('Matrice di Correlazione delle Variabili', fontsize=16, pad=20)

    # 7. Ottimizzazione della leggibilità delle etichette degli assi.
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)

    # 8. Layout compatto per evitare sovrapposizioni.
    fig.tight_layout()

    # 9. Salvataggio dell'immagine in un file PNG.
    plt.savefig('correlation_matrix_enhanced.png')
    
    print("Figura 'correlation_matrix_enhanced.png' salvata correttamente.")


except FileNotFoundError:
    print("Errore: Il file 'AR_DR_results.csv' non è stato trovato.")
except Exception as e:
    print(f"Si è verificato un errore imprevisto: {e}")
```
