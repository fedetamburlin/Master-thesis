# Sorting

##### Per agevolare il post processing

```python
import pandas as pd
import numpy as np
import os

# Cartelle di input e output
input_folder = "extracted_data"
output_folder = "sorted_data"

# Assicurati che la cartella di output esista
os.makedirs(output_folder, exist_ok=True)


# Funzione per ordinare i punti in senso antiorario
def order_points_ccw(df):
    
    # Rimuovi duplicati nel numero Nodo (prima colonna)
    df = df.drop_duplicates(subset=["Nodo"])

    # Funzione per ordinare un gruppo di punti per Z
    def sort_group(group):
        centroid_x, centroid_y = group["X"].mean(), group["Y"].mean()
        group["angle"] = np.arctan2(group["Y"] - centroid_y, group["X"] - centroid_x)
        return group.sort_values("angle", ascending=False).drop(columns="angle")

    return df.groupby("Z", group_keys=False).apply(sort_group)


# Processa tutti i file CSV nella cartella di input
for filename in os.listdir(input_folder):
    if filename.endswith(".csv"):
        input_path = os.path.join(input_folder, filename)
        output_path = os.path.join(output_folder, filename)

        # Carica il file CSV
        df = pd.read_csv(input_path)

        # Ordina i punti e rimuove duplicati
        df_sorted = order_points_ccw(df)

        # Salva il file ordinato
        df_sorted.to_csv(output_path, index=False)


print("Operazione completata! I file ordinati e puliti sono stati salvati in 'sorted_data'.")
```
